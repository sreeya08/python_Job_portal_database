# Job Portal Management System
## Python + TXT File Project
This project uses only Python. Data is stored in `job_portal_data.txt`. No SQL or database is used.

```python
DATA_FILE = 'job_portal_data.txt'
```

```python
def load_data():
    data = {}
    section = None
    with open(DATA_FILE, "r", encoding="utf-8") as file:
        for line in file:
            line = line.strip()
            if not line:
                continue
            if line.startswith("[") and line.endswith("]"):
                section = line[1:-1]
                data[section] = []
            else:
                data[section].append(line.split("|"))
    return data

data = load_data()
print("Data loaded successfully!")
print("Modules:", ", ".join(data.keys()))
```

```python
def view_candidates():
    print("\n--- CANDIDATES ---")
    for c in data["CANDIDATES"]:
        print(f"ID: {c[0]} | Name: {c[1]} | Qualification: {c[4]} | Experience: {c[5]} | City: {c[6]}")

def view_companies():
    print("\n--- COMPANIES ---")
    for c in data["COMPANIES"]:
        print(f"ID: {c[0]} | Company: {c[1]} | Industry: {c[2]} | Location: {c[3]}")

def view_jobs():
    print("\n--- JOBS ---")
    for j in data["JOBS"]:
        print(f"ID: {j[0]} | Job: {j[2]} | Salary: ₹{j[3]} | Location: {j[4]}")

def view_skills():
    print("\n--- SKILLS ---")
    for s in data["SKILLS"]:
        print(f"ID: {s[0]} | Skill: {s[1]}")
```

```python
def search_jobs():
    location = input("Enter location: ").strip().lower()
    found = False
    for job in data["JOBS"]:
        if job[4].lower() == location:
            print(f"ID: {job[0]} | {job[2]} | ₹{job[3]} | {job[4]}")
            found = True
    if not found:
        print("No jobs found.")

def apply_for_job():
    candidate_id = input("Enter candidate ID: ").strip()
    job_id = input("Enter job ID: ").strip()

    if not any(c[0] == candidate_id for c in data["CANDIDATES"]):
        print("Candidate ID not found.")
        return
    if not any(j[0] == job_id for j in data["JOBS"]):
        print("Job ID not found.")
        return

    next_id = max(int(a[0]) for a in data["APPLICATIONS"]) + 1
    data["APPLICATIONS"].append([str(next_id), candidate_id, job_id, "2026-08-31", "Pending"])
    print(f"Application submitted successfully! Application ID: {next_id}")
```

```python
def application_status():
    candidate_id = input("Enter candidate ID: ").strip()
    for app in data["APPLICATIONS"]:
        if app[1] == candidate_id:
            job = next(j for j in data["JOBS"] if j[0] == app[2])
            print(f"Job: {job[2]} | Date: {app[3]} | Status: {app[4]}")

def hiring_report():
    counts = {}
    for app in data["APPLICATIONS"]:
        counts[app[4]] = counts.get(app[4], 0) + 1
    print("\n--- HIRING REPORT ---")
    for status, total in counts.items():
        print(f"{status}: {total}")

def interview_report():
    print("\n--- INTERVIEW REPORT ---")
    for i in data["INTERVIEWS"]:
        app = next(a for a in data["APPLICATIONS"] if a[0] == i[1])
        candidate = next(c for c in data["CANDIDATES"] if c[0] == app[1])
        job = next(j for j in data["JOBS"] if j[0] == app[2])
        print(f"{candidate[1]} | {job[2]} | {i[2]} | {i[3]}")
```

```python
def main():
    while True:
        print("\n===== JOB PORTAL MANAGEMENT SYSTEM =====")
        print("1. View Candidates")
        print("2. View Companies")
        print("3. View Jobs")
        print("4. View Skills")
        print("5. Search Jobs")
        print("6. Apply for Job")
        print("7. Check Application Status")
        print("8. Hiring Report")
        print("9. Interview Report")
        print("10. Exit")

        choice = input("Enter your choice: ")

        if choice == "1": view_candidates()
        elif choice == "2": view_companies()
        elif choice == "3": view_jobs()
        elif choice == "4": view_skills()
        elif choice == "5": search_jobs()
        elif choice == "6": apply_for_job()
        elif choice == "7": application_status()
        elif choice == "8": hiring_report()
        elif choice == "9": interview_report()
        elif choice == "10":
            print("Thank you!")
            break
        else:
            print("Invalid choice.")

main()
```
